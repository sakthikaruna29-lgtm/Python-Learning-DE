```python
A1=[1,2,3,5,10,"Ben",12.3,982332,'d',[133,324,244],(13,"ben",19)]

# append() -> Adds the ENTIRE argument as a SINGLE element at the end of the list
# (even if the argument is itself a list, it goes in as one nested item, not merged)
A1.append([12,12,[10,12]])

print(A1)
# Output:
# [1, 2, 3, 5, 10, 'Ben', 12.3, 982332, 'd', [133, 324, 244], (13, 'ben', 19), [12, 12, [10, 12]]]


# extend() -> Adds each element of the given iterable individually to the end of the list
# (unlike append, it "unpacks" the argument and merges its items one by one)
A1.extend([10,12])

print("Extend" , A1)
# Output:
# Extend [1, 2, 3, 5, 10, 'Ben', 12.3, 982332, 'd', [133, 324, 244], (13, 'ben', 19), [12, 12, [10, 12]], 10, 12]


# pop() -> Removes and returns the LAST element of the list (default index = -1)
A1.pop()
print("Pop" , A1)
# Removes 12 (the last element added by extend)
# Output:
# Pop [1, 2, 3, 5, 10, 'Ben', 12.3, 982332, 'd', [133, 324, 244], (13, 'ben', 19), [12, 12, [10, 12]], 10]


# pop(index) -> Removes and returns the element at the SPECIFIED index
A1.pop(3)
print("Pop()" , A1) # will remove 5 from the list
# Index 3 currently holds value 5, so it gets removed
# Output:
# Pop() [1, 2, 3, 10, 'Ben', 12.3, 982332, 'd', [133, 324, 244], (13, 'ben', 19), [12, 12, [10, 12]], 10]


# remove(value) -> Removes the FIRST OCCURRENCE of the given VALUE (not index)
# Raises ValueError if the value is not found
A1.remove(10)
print("Remove" , A1)
# The first 10 in the list (right after 2, 3) gets removed
# Note: the second 10 at the end of the list is untouched
# Output:
# Remove [1, 2, 3, 'Ben', 12.3, 982332, 'd', [133, 324, 244], (13, 'ben', 19), [12, 12, [10, 12]], 10]


# reverse() -> Reverses the order of elements IN PLACE (no new list created)
A1.reverse()
print("Reverse" , A1)
# Output:
# Reverse [10, [12, 12, [10, 12]], (13, 'ben', 19), [133, 324, 244], 'd', 982332, 12.3, 'Ben', 3, 2, 1]


'''A1.sort()
print("Sort" , A1) # If list have multiple data type Sort wont work it will throw errior '''
# sort() -> Sorts elements IN PLACE in ascending order by default
# It compares elements pairwise using < , which requires all elements
# to be of COMPARABLE/COMPATIBLE types (e.g., all int, all str, etc.)
# Since A1 has mixed types (int, str, float, tuple, list),
# Python cannot compare e.g. int < str or int < list
# => Raises: TypeError: '<' not supported between instances of 'str' and 'int' (or similar)


a2=[234,434,1000000,1,0.9,23443,2]

# sort() -> Works fine here since ALL elements are numeric (int/float),
# so they are mutually comparable. Sorts ascending in place.
a2.sort()
print("Sort" , a2)
# Output:
# Sort [0.9, 1, 2, 234, 434, 23443, 1000000]


# clear() -> Removes ALL elements from the list, leaving it empty ([])# Creating a set
# A set is an UNORDERED collection of UNIQUE elements (no duplicates allowed)
# NOTE: Since sets are unordered, the printed order of elements can vary
# between runs/machines — the outputs below show ONE possible ordering
my_set = {1, 2, 3, 5, 10, "Ben", 12.3}
print("Original Set:", my_set)
# Output (order may vary):
# Original Set: {1, 2, 3, 5, 10, 12.3, 'Ben'}


# add() -> Adds a SINGLE element to the set
# If the element already exists, nothing happens (no error, no duplicate)
my_set.add(15)
print("After add(15):", my_set)
# Output:
# After add(15): {1, 2, 3, 5, 10, 12.3, 15, 'Ben'}


# update() -> Adds MULTIPLE elements (from any iterable) to the set
# Duplicates (already-existing elements) are automatically ignored
my_set.update([20, 25, 10]) # Adding multiple elements, 10 is ignored as it's a duplicate
print("After update([20, 25, 10]):", my_set)
# Output:
# After update([20, 25, 10]): {1, 2, 3, 5, 10, 12.3, 15, 20, 25, 'Ben'}


# remove(value) -> Removes the specified value from the set
# Raises KeyError if the value does NOT exist in the set
my_set.remove("Ben") # Raises KeyError if element not found
print("After remove('Ben'):", my_set)
# Output:
# After remove('Ben'): {1, 2, 3, 5, 10, 12.3, 15, 20, 25}


# discard(value) -> Removes the specified value from the set
# UNLIKE remove(), does NOT raise an error if the value is not found
# (safer choice when you're not sure the element exists)
my_set.discard(100) # Does nothing if element not found
print("After discard(100):", my_set)
# Output (unchanged, since 100 wasn't present):
# After discard(100): {1, 2, 3, 5, 10, 12.3, 15, 20, 25}


# pop() -> Removes and returns an ARBITRARY element from the set
# Since sets are unordered, you CANNOT predict which element will be removed
# Raises KeyError if the set is empty
popped_element = my_set.pop() # Removes and returns an arbitrary element
print(f"After pop() (removed {popped_element}):", my_set)
# Output (element removed is arbitrary — commonly the "smallest hashed" one, here shown as 1):
# After pop() (removed 1): {2, 3, 5, 10, 12.3, 15, 20, 25}


# Set operations
set_b = {5, 15, 30, "Alice"}
print("Set B:", set_b)
# Output:
# Set B: {5, 15, 30, 'Alice'}


# union() -> Returns a NEW set containing ALL elements from BOTH sets
# (equivalent to using the | operator) — does NOT modify original sets
print("Union (my_set | set_b):", my_set.union(set_b))
# Output:
# Union (my_set | set_b): {2, 3, 5, 10, 12.3, 15, 20, 25, 30, 'Alice'}


# intersection() -> Returns a NEW set containing ONLY elements COMMON to BOTH sets
# (equivalent to using the & operator)
print("Intersection (my_set & set_b):", my_set.intersection(set_b))
# Output:
# Intersection (my_set & set_b): {5, 15}


# difference() -> Returns a NEW set containing elements that are in my_set
# but NOT in set_b (order of operands matters!)
# (equivalent to using the - operator)
print("Difference (my_set - set_b):", my_set.difference(set_b))
# Output:
# Difference (my_set - set_b): {2, 3, 10, 12.3, 20, 25}


# Clearing a set
# clear() -> Removes ALL elements, leaving an empty set: set()
my_set.clear()
print("After clear():", my_set)
# Output:
# After clear(): set()
a2.clear()
print("Clear" , a2)
# Output:
# Clear []
```

```python
# Creating a set
# A set is an UNORDERED collection of UNIQUE elements (no duplicates allowed)
# NOTE: Since sets are unordered, the printed order of elements can vary
# between runs/machines — the outputs below show ONE possible ordering
my_set = {1, 2, 3, 5, 10, "Ben", 12.3}
print("Original Set:", my_set)
# Output (order may vary):
# Original Set: {1, 2, 3, 5, 10, 12.3, 'Ben'}


# add() -> Adds a SINGLE element to the set
# If the element already exists, nothing happens (no error, no duplicate)
my_set.add(15)
print("After add(15):", my_set)
# Output:
# After add(15): {1, 2, 3, 5, 10, 12.3, 15, 'Ben'}


# update() -> Adds MULTIPLE elements (from any iterable) to the set
# Duplicates (already-existing elements) are automatically ignored
my_set.update([20, 25, 10]) # Adding multiple elements, 10 is ignored as it's a duplicate
print("After update([20, 25, 10]):", my_set)
# Output:
# After update([20, 25, 10]): {1, 2, 3, 5, 10, 12.3, 15, 20, 25, 'Ben'}


# remove(value) -> Removes the specified value from the set
# Raises KeyError if the value does NOT exist in the set
my_set.remove("Ben") # Raises KeyError if element not found
print("After remove('Ben'):", my_set)
# Output:
# After remove('Ben'): {1, 2, 3, 5, 10, 12.3, 15, 20, 25}


# discard(value) -> Removes the specified value from the set
# UNLIKE remove(), does NOT raise an error if the value is not found
# (safer choice when you're not sure the element exists)
my_set.discard(100) # Does nothing if element not found
print("After discard(100):", my_set)
# Output (unchanged, since 100 wasn't present):
# After discard(100): {1, 2, 3, 5, 10, 12.3, 15, 20, 25}


# pop() -> Removes and returns an ARBITRARY element from the set
# Since sets are unordered, you CANNOT predict which element will be removed
# Raises KeyError if the set is empty
popped_element = my_set.pop() # Removes and returns an arbitrary element
print(f"After pop() (removed {popped_element}):", my_set)
# Output (element removed is arbitrary — commonly the "smallest hashed" one, here shown as 1):
# After pop() (removed 1): {2, 3, 5, 10, 12.3, 15, 20, 25}


# Set operations
set_b = {5, 15, 30, "Alice"}
print("Set B:", set_b)
# Output:
# Set B: {5, 15, 30, 'Alice'}


# union() -> Returns a NEW set containing ALL elements from BOTH sets
# (equivalent to using the | operator) — does NOT modify original sets
print("Union (my_set | set_b):", my_set.union(set_b))
# Output:
# Union (my_set | set_b): {2, 3, 5, 10, 12.3, 15, 20, 25, 30, 'Alice'}


# intersection() -> Returns a NEW set containing ONLY elements COMMON to BOTH sets
# (equivalent to using the & operator)
print("Intersection (my_set & set_b):", my_set.intersection(set_b))
# Output:
# Intersection (my_set & set_b): {5, 15}


# difference() -> Returns a NEW set containing elements that are in my_set
# but NOT in set_b (order of operands matters!)
# (equivalent to using the - operator)
print("Difference (my_set - set_b):", my_set.difference(set_b))
# Output:
# Difference (my_set - set_b): {2, 3, 10, 12.3, 20, 25}


# Clearing a set
# clear() -> Removes ALL elements, leaving an empty set: set()
my_set.clear()
print("After clear():", my_set)
# Output:
# After clear(): set()
```

## Dictionary Examples

```python
# Creating a dictionary
# A dictionary stores data as KEY-VALUE pairs
# Keys must be UNIQUE and HASHABLE (immutable), values can be anything
# From Python 3.7+, dictionaries maintain INSERTION ORDER
my_dict = {
    "name": "Alice",
    "age": 30,
    "city": "New York",
    "occupation": "Engineer"
}
print("Original Dictionary:", my_dict)
# Output:
# Original Dictionary: {'name': 'Alice', 'age': 30, 'city': 'New York', 'occupation': 'Engineer'}


# Accessing values

# Using [] -> Directly fetches the value for the given key
# Raises KeyError if the key does NOT exist
print("Name:", my_dict["name"])
# Output:
# Name: Alice


# get() -> Fetches the value for the given key
# Returns None (or a default you specify) if key not found — NO error raised
print("Age (using .get()):", my_dict.get("age"))
# Output:
# Age (using .get()): 30


# get(key, default) -> Same as above, but returns the SPECIFIED default
# instead of None when the key is missing
print("Country (non-existent key using .get()):", my_dict.get("country", "Unknown"))
# Output:
# Country (non-existent key using .get()): Unknown


# Adding and modifying key-value pairs

# dict[new_key] = value -> Adds a NEW key-value pair if key doesn't exist
my_dict["email"] = "alice@example.com"
print("After adding email:", my_dict)
# Output:
# After adding email: {'name': 'Alice', 'age': 30, 'city': 'New York', 'occupation': 'Engineer', 'email': 'alice@example.com'}


# dict[existing_key] = new_value -> UPDATES the value if key already exists
my_dict["age"] = 31 # Modifying an existing value
print("After modifying age:", my_dict)
# Output:
# After modifying age: {'name': 'Alice', 'age': 31, 'city': 'New York', 'occupation': 'Engineer', 'email': 'alice@example.com'}


# Removing key-value pairs

# pop(key) -> Removes the key-value pair for the given KEY and RETURNS the value
# Raises KeyError if key not found (unless a default is given: pop(key, default))
removed_occupation = my_dict.pop("occupation")
print(f"After pop('occupation') (removed {removed_occupation}):", my_dict)
# Output:
# After pop('occupation') (removed Engineer): {'name': 'Alice', 'age': 31, 'city': 'New York', 'email': 'alice@example.com'}


# popitem() -> Removes and returns the LAST INSERTED key-value pair as a tuple (key, value)
# In Python 3.7+, dicts are ordered, so this always removes the most RECENTLY ADDED item
# Raises KeyError if the dictionary is empty
last_item = my_dict.popitem() # Removes and returns the last inserted item (Python 3.7+)
print(f"After popitem()
```

```python
# Creating a tuple
# A tuple is an ORDERED, IMMUTABLE collection (cannot be changed after creation)
# Can hold DUPLICATE values and MIXED data types
my_tuple = (1, 2, 3, 5, 10, "Ben", 12.3, 5)
print("Original Tuple:", my_tuple)
# Output:
# Original Tuple: (1, 2, 3, 5, 10, 'Ben', 12.3, 5)


# Single-element tuple -> MUST include a trailing comma
# Without the comma, Python treats it as just a regular value in parentheses (NOT a tuple)
single_tuple = (5,)
not_a_tuple = (5)
print("Single-element tuple:", single_tuple, type(single_tuple))
print("Not a tuple:", not_a_tuple, type(not_a_tuple))
# Output:
# Single-element tuple: (5,) <class 'tuple'>
# Not a tuple: 5 <class 'int'>


# Accessing elements

# Indexing -> Access element at a specific position (0-based)
print("Element at index 2:", my_tuple[2])
# Output:
# Element at index 2: 3


# Negative indexing -> Access from the END of the tuple
print("Last element:", my_tuple[-1])
# Output:
# Last element: 5


# Slicing -> Returns a NEW tuple with a subset of elements [start:stop]
print("Slice [1:4]:", my_tuple[1:4])
# Output:
# Slice [1:4]: (2, 3, 5)


# Tuple methods (only 2 exist, since tuples are immutable)

# count(value) -> Returns the NUMBER of times a value appears in the tuple
print("Count of 5:", my_tuple.count(5))
# Output:
# Count of 5: 2


# index(value) -> Returns the INDEX of the FIRST OCCURRENCE of a value
# Raises ValueError if the value is not found
print("Index of 'Ben':", my_tuple.index("Ben"))
# Output:
# Index of 'Ben': 5


# Immutability demonstration

# Tuples CANNOT be modified after creation — no append/remove/sort/etc.
# Attempting item assignment raises a TypeError
try:
    my_tuple[0] = 100
except TypeError as e:
    print("Error when modifying tuple:", e)
# Output:
# Error when modifying tuple: 'tuple' object does not support item assignment


# Concatenation -> Combines two tuples into a NEW tuple using +
# (does NOT modify the original tuples)
tuple_b = (20, 25)
combined_tuple = my_tuple + tuple_b
print("Concatenated Tuple:", combined_tuple)
# Output:
# Concatenated Tuple: (1, 2, 3, 5, 10, 'Ben', 12.3, 5, 20, 25)


# Repetition -> Repeats the tuple's elements N times using *
repeated_tuple = tuple_b * 3
print("Repeated Tuple:", repeated_tuple)
# Output:
# Repeated Tuple: (20, 25, 20, 25, 20, 25)


# Membership check -> Checks if a value EXISTS in the tuple (True/False)
print("Is 10 in my_tuple?", 10 in my_tuple)
# Output:
# Is 10 in my_tuple? True


# Length -> Returns the NUMBER of elements in the tuple
print("Length of my_tuple:", len(my_tuple))
# Output:
# Length of my_tuple: 8


# Unpacking -> Assigns tuple elements to individual variables in one line
# Number of variables must match number of elements (unless using * for extras)
a, b, c = (10, 20, 30)
print("Unpacked values:", a, b, c)
# Output:
# Unpacked values: 10 20 30


# Converting to a list and back (common workaround since tuples are immutable)
# If you truly need to "modify" a tuple, convert to list, change it, convert back
temp_list = list(my_tuple)
temp_list[0] = 100
my_tuple = tuple(temp_list)
print("Tuple after workaround modification:", my_tuple)
# Output:
# Tuple after workaround modification: (100, 2, 3, 5, 10, 'Ben', 12.3, 5)
```
