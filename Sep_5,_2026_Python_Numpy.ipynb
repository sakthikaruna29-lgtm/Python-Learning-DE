```python
import numpy as np
```

```python
# Array Creation
```

```python
arr=np.array([1,2,3])
print (arr)
arr.ndim
```

```python
a=np.array([[1,2,3],[1,2,3]])
print(a)
a.ndim
```

```python
b=np.array([[[1,2,3],[1,2,3]],[[1,5,4],[1,4,5]]])
print(b)
b.ndim
```

```python
# Array Creation Methods
```

```python
c=np.array([1,2,3])
d=np.zeros(5)
e=np.ones(4,dtype=int) # Print 1D array with 1 as value for 4 columns
f=np.arange(1,10) #1 to 9
g=np.linspace(1,10,3) # 3 evenly spaced value

print(c)
print(d)
print(e)
print(f)
print(g)
```

```python
# Shapes and Dimensions Methods
```

```python
fl=np.array([[19,3,134],[2,43,22]])
print(fl.shape)
print(fl.ndim)
print(fl.size)
print(fl.reshape(6,1))
print(fl.flatten())
```

```python
# Mathematical Methods
```

```python
print(fl.sum())
print(fl.mean())
print(fl.max())
print(fl.min())
print(fl.std())
print(fl.var())
```

```python
# Sorting Methods
```

```python
print(np.sort(fl))
print(np.argsort(fl))
```

```python
# Joining and Splitting Arrays
```

```python
a=np.array([[1,2,3],[3,4,5]])
b=np.array([[8,3,2],[4,6,5]])

print("Concatenate:", np.concatenate((a,b)))
print("vstack", np.vstack((a,b)))
print("hstack",np.hstack((a,b)))
```
