```python
# List Comprehension

# Create a list using list comprehension syntax: [expression for item in iterable if condition]
# - i                       -> the expression: just keep the value of i as-is (no transformation)
# - for i in range(0, 11)   -> loop i from 0 to 10 (11 is exclusive, so range stops at 10)
# - if i % 2 == 0           -> filter: only include i when it's even (remainder 0 when divided by 2)
a = [i for i in range(0, 11) if i % 2 == 0]

# Print the resulting list
print(a)
```

```python
# Dict Comprehension

# Create a dictionary using dict comprehension syntax: {key: value for item in iterable if condition}
# - i : i        -> key is i, value is also i (so key == value)
# - for i in range(0, 11)  -> loop i from 0 to 10 (11 is exclusive)
# - if i % 2 == 0           -> only include i if it's divisible by 2 (i.e., even numbers)
a2 = {i: i for i in range(0, 11) if i % 2 == 0}

# Print the resulting dictionary
print(a2)
```

```python
marks={"Sakthi":87,"Ram":98,"Sanjay":67,"Sai":30,"Bunny":10}
rank1={k:v for k,v in marks.items() if v >=80}
print(rank1)

rank2={k:v for k,v in marks.items() if v>=50 and v<80}
print(rank2)

rank3={k:v for k,v in marks.items() if v>=30 and v<50}
print(rank3)

rank4={k:v for k,v in marks.items() if  v<30}
print(rank4, "Fail :( ")
```

```python
# Dict comprehension with if-else (ternary) — labeling values without dropping any keys

# Original dict: student names mapped to their marks
marks1 = {"Sakthi": 87, "Ram": 98, "Sanjay": 67, "Sai": 30, "Bunny": 10}

# Build a new dict from marks1:
# - k : the key (student name) — copied over unchanged
# - v if v >= 30 else "Fail"  -> the value:
#       if marks (v) are 30 or above, keep the mark as-is
#       otherwise, replace it with the string "Fail"
# - for k, v in marks1.items() -> loop through each (name, mark) pair
# NOTE: no filter `if` after the for-loop, so ALL keys stay in the result —
#       this is the if-else (ternary) pattern, not the filter pattern
rank5 = {k: (v if v >= 30 else "Fail") for k, v in marks1.items()}

# Print the resulting dict
print(rank5)
```

```python
products={"Mobile":70000,"Fridge":27000,"washing Machine":30000}

Price_Including_GST={k:v+(v*0.18) for k,v in products.items()}
# Build a new dict from `products`:
# - k                  -> the key (product name), copied over unchanged
# - v + (v * 0.18)     -> the value: base price (v) plus 18% of the base price (v * 0.18)
#                          this gives the final price after adding GST
# - for k, v in products.items()  -> loop through each (product, price) pair
print(Price_Including_GST)
```
