```python
# Find the largest among three numbers.

a=int(input())
b=int(input())
c=int(input())

if a>b and a>c:
  print(f"{a} is greater than {b} and {c}")
elif b>a and b>c:
  print(f"{b} is greater than {a} and {c}")
else:
  print(f"{c} is greater than {a} and {b}")
```

```python
# Check whether a year is a leap year.
Year=int(input("Enter the year:"))

if (Year%4==0) and (Year%100!=0):
  print(f"Year {Year} is a Leap year")
else:
  print(f"Year {Year} is not a Leap Year")
```

```python
# Calculate simple interest.

Principal=float(input("Enter the principal amt:"))
Intrest_rate=int(input("Enter the intrest rate:"))
Time_years=int(input("Enter the time in years:"))

Simple_intrest = (Principal*Intrest_rate*Time_years)/100

print(f"Simple Intrest is {Simple_intrest}")
```

```python
"""Grade a student based on marks:
90+ → A
75–89 → B
50–74 → C
Below 50 → Fail
"""
Tam=float(input("Tamil:"))
Eng=float(input("English:"))
Mat=float(input("Maths:"))
Sci=float(input("Science:"))
Soc=float(input("Social:"))

Avg=(Tam+Eng+Mat+Sci+Soc)/5

print(Avg)

if Avg>=90:
  print("Grade is A")
elif Avg>=75 and Avg<=89:
  print("Grade is B")
elif Avg>=50 and Avg<=74:
  print("Grade is C")
else:
  print("Fail")
```

```python

# Create a calculator using if-elif.
print("1.Addition")
print("2.Subraction")
print("3.Multiplication")
print("4.Division")

Num1=float(input())
Num2=float(input())
Cal=int(input("Enter the calculation:"))

if Cal==1:
  Num3=Num1+Num2
  print(Num3)
elif Cal==2:
  Num3=Num1-Num2
  print(Num3)
elif Cal==3:
  Num3=Num1*Num2
  print(Num3)
elif Cal==4:
  Num3=Num1/Num2
  print(Num3)
else:
  print("Invalid input")

```

```python
# Find whether a number is divisible by both 3 and 5.

Num=int(input("Enter the Number:"))

if Num%3==0 and Num%5==0:
  print("Number is divisible by 3 and 5")
elif Num%3==0 and Num%5!=0:
  print("Number is not divisible by 5")
elif Num%3!=0 and Num%5==0:
  print("Number is not divisible by 3")
```
